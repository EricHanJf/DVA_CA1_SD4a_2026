# Theoretical Framework

A composite index is used to measure a multidimensional concept by combining several related indicators into a single overall score. In this project, the concept being measured is the **cost of international education** at country level.

The cost of studying abroad cannot be represented accurately by a single variable. Although tuition fees are often one of the main expenses for international students, they are not the only cost involved. Students also need to consider accommodation, day-to-day living expenses, visa costs, and insurance charges. As a result, two countries may appear similar in terms of tuition fees but differ significantly in their overall cost once other expenses are taken into account. For this reason, a composite index is a more appropriate tool for comparing the financial burden of international education across countries.

In this project, the index is structured around three main dimensions.
### 2.1 Tuition Cost
This dimension represents the direct academic cost of studying in a country. Tuition fees usually make up the largest single cost for international students and therefore form an essential part of the index. 
- **Tuition_USD**

### 2.2 Living Cost
This dimension represents the general cost of living associated with studying in a country. It includes both overall living expenses and accommodation-related cost.
- **Living_Cost_Index**
- **Rent_USD**

These variables are important because even when tuition fees are relatively affordable, high living or housing costs can still make a country expensive overall.

### 2.3 Additional Mandatory Costs
This dimension captures extra costs that international students are often required to pay in addition to tuition and living expenses.
- **Visa_Fee**
- **Insurance_USD**

Although these costs may be smaller than tuition or rent, they still contribute to the total financial burden and should be included in the index.

### 2.4 Variables not included in the final index
Some variables in the dataset are useful for context, description, or grouping, but are not included directly in the final composite index.
- **Country** is used as the main unit of comparison rather than as an indicator.
- **City**, **University**, **Program**, and **Level** are descriptive variables and do not directly measure cost at country level.
- **Duration** may be relevant when estimating total full-programme cost, but this project focuses on comparing cost indicators directly rather than calculating the full cost of an entire degree.
- **Exchange_Rate** is not included in the final index because the main monetary variables are already expressed in USD, which makes direct comparison possible without using exchange rate as a separate indicator.

Overall, this framework treats international education cost as a multidimensional concept made up of tuition cost, living cost, and additional mandatory expenses. This structure provides the basis for selecting indicators, aggregating the data at country level, and constructing the final composite index.
