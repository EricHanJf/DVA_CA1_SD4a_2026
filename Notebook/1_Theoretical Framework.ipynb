# Sustainable Living Quality Index

## 1.1 Introduce of The Index
The aim of this project is to develop a Sustainable Living Quality Index (SLQI) to compare countries based on their overall quality of life and sustainability. The index is designed to measure more than economic wealth alone. Although GDP per capita is an important indicator, it cannot fully explain whether a country provides good living conditions for its population.

For this reason, the SLQI uses a multidimensional framework. It combines economic, social, infrastructure, environmental and governance indicators into one composite index. This allows countries to be compared using a broader and more balanced view of sustainable living quality.

## 1.2 Conceptual Model / Dimensions
The theoretical framework is based on five dimensions: economic stability, health and education, basic infrastructure, environmental sustainability, and governance stability. These dimensions were selected because they represent key areas that influence daily living conditions and long-term national development.

Economic stability reflects whether people live in a stable economic environment. Health and education represent human development and future opportunities. Basic infrastructure measures access to essential services such as electricity, drinking water and sanitation. Environmental sustainability reflects the quality of the natural environment and the use of cleaner energy. Governance stability represents the ability of institutions to provide effective public services, rule of law and political stability.
