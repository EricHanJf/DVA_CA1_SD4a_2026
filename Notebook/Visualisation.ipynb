## Visualisation

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load final SLQI result
slqi_df = pd.read_csv("../Dataset/final_SLQI_index_score.csv")

# Select top 20 countries
top_20 = slqi_df.sort_values("slqi_rank").head(20)

plt.figure(figsize=(10, 12))
plt.barh(top_20["Country"], top_20["slqi_score"])
plt.xlabel("SLQI Score")
plt.ylabel("Country")
plt.title("Top 20 Countries by SLQI Score")

# Highest rank at the top
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

This chart presents the top 50 countries based on the final SLQI score. Countries with higher scores show stronger overall sustainable living quality across the selected dimensions.

```python
bottom_20 = slqi_df.sort_values("slqi_rank", ascending=False).head(20)
plt.figure(figsize=(10, 13))
plt.barh(bottom_20["Country"],bottom_20["slqi_score"])
plt.xlabel("SLQI Score")
plt.ylabel("Country")
plt.title("Bottom 20 Countries by SLQI Score")

# Highest rank at the top
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

This chart shows the lowest-ranked countries in the SLQI. These countries generally have lower performance across several dimensions, especially infrastructure, health and education, or governance.

```python
dimension_cols = ["economic_score", "health_education_score", "infrastructure_score", "governance_score", "environment_score"]
top_20_dimension = top_20.set_index("Country")[dimension_cols]
top_20_dimension.plot(kind="bar", figsize=(14, 7))
plt.ylabel("Dimension Score")
plt.xlabel("Country")
plt.title("Dimension Scores of Top 20 SLQI Countries")
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```
