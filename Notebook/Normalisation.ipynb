## Normalisation

## 1. loading the cleaned dataset

```python
import pandas as pd
import numpy as np

slqi_dataset = pd.read_csv("../Dataset/final_SLQI_dataset.csv")
slqi_dataset.head()
```

## 2. Select indicators

```python
indicator_cols = [
    "gdp_per_capita",
    "inflation",
    "unemployment_rate",
    "life_expectancy",
    "secondary_enrollment",
    "electricity_access",
    "drinking_water_access",
    "sanitation_access",
    "government_effectiveness",
    "political_stability",
    "rule_of_law",
    "forest_area",
    "pm25_exposure",
    "renewable_energy"
]
```

## 3. Define positive and negative indicators

```python
positive_indicators = [
    "gdp_per_capita",
    "life_expectancy",
    "secondary_enrollment",
    "electricity_access",
    "drinking_water_access",
    "sanitation_access",
    "government_effectiveness",
    "political_stability",
    "rule_of_law",
    "forest_area",
    "renewable_energy"
]
```

```python
negative_indicators = [
    "inflation",
    "unemployment_rate",
    "pm25_exposure"
]
```

## 4. Apply Min-Max Normalisation

```python
slqi_normalised_dataset = slqi_dataset.copy()
```

Positive indicators formula
(value - min) / (max - min)
Negative indicators formula
(max - value) / (max - min)

```python
# Normalise positive indicators
for col in positive_indicators:
    min_value = slqi_normalised_dataset[col].min()
    max_value = slqi_normalised_dataset[col].max()
    
    slqi_normalised_dataset[col + "_normalised"] = (
        (slqi_normalised_dataset[col] - min_value) / (max_value - min_value)
    )

# Normalise negative indicators
for col in negative_indicators:
    min_value = slqi_normalised_dataset[col].min()
    max_value = slqi_normalised_dataset[col].max()
    
    slqi_normalised_dataset[col + "_normalised"] = (
        (max_value - slqi_normalised_dataset[col]) / (max_value - min_value)
    )
```

## 5. Check normalised values

```python
normalised_cols = [col + "_normalised" for col in indicator_cols]
slqi_normalised_dataset[normalised_cols].head()
```

```python
slqi_normalised_dataset[normalised_cols].describe()
```

```python
slqi_normalised_dataset.to_csv("../Dataset/normalised_SLQI_dataset.csv", index=False)
```
